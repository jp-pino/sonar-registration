```python
import gzip

import delimited_protobuf
from protocol import message_formats_pb2, telemetry_pb2
import numpy as np

from utils import binlog
```

```python
def read_binlog(file_path):
    with gzip.open(file_path, "rb") as fin:
        
        while True:
            try:
                binlog = delimited_protobuf.read(fin, message_formats_pb2.BinlogRecord)
                # print(binlog)
                if (
                        telemetry_pb2.PositionEstimateTel.DESCRIPTOR.full_name
                        not in binlog.payload.type_url
                ):
                    continue
                imu = telemetry_pb2.PositionEstimateTel()
                imu.ParseFromString(binlog.payload.value)
                yield imu, binlog.unix_timestamp.seconds + binlog.unix_timestamp.nanos / 1e9
            except Exception as e:
                print(e)
                return
```

```python
# Open generator
generator = read_binlog("logs/BYEDP210001_ee6c4d5b1c0c71d4_01077.bez")

# Read the first ping
_, prev_ts = next(generator)
for i in range(10):
    pos, ts = next(generator)
    
    print(pos)
    # print(ts)
    
    time = ts - prev_ts
    print(time)
    
    prev_ts = ts
    
    
```
